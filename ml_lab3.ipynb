```python

def minkowski_distance(x, y, p):
  total = 0
  for i in range(len(x)):
    total += abs(x[i] - y[i]) ** p
    return total ** (1 / p)
a = [2, 3, 4]
b = [5, 7, 8]

print("Manhattan:", minkowski_distance(a, b, 1))
print("Euclidean:", minkowski_distance(a, b, 2))
```

```python
import pandas as pd


df = pd.read_excel("/content/Lab Session Data.xlsx", sheet_name= "marketing_campaign")



def label_encode(column):
    unique_values = column.unique()
    mapping = {}

    for i, value in enumerate(unique_values):
        mapping[value] = i

    encoded = column.map(mapping)
    return encoded, mapping




def one_hot_encode(column):
    unique_values = column.unique()

    encoded_df = pd.DataFrame()

    for value in unique_values:
        encoded_df[value] = (column == value).astype(int)

    return encoded_df

df["Education"], education_mapping = label_encode(df["Education"])

print("Education Mapping:")
print(education_mapping)




marital_encoded = one_hot_encode(df["Marital_Status"])

df = df.drop("Marital_Status", axis=1)

df = pd.concat([df, marital_encoded], axis=1)



print("\nDataset Shape After Encoding:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nFirst Five Rows:")
print(df.head())
```
